```python
import pandas as pd

info1 = {
    "Name": ["Adam", "Eve", "Bob"],
    "Age": [23, 24, 25],
    "GPA": [9.5, 8.6, 7.2]
}

df1 = pd.DataFrame(info1)

print(df1, type(df1))

print(df1.index)
print(df1.columns)
```

```python
df = pd.DataFrame(
    [["Adam", 23], ["Bob", 23], ["Eve", 23]],
    columns=["Name", "Age"]
)

print(df)
```

```python
import numpy as np

np_arr = np.array([[1, 2, 3], [4, 5, 6]])

df = pd.DataFrame(np_arr, columns=["A", "B", "C"])

print(df)
```
